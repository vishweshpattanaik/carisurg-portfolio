## Week 0 Assignment 2

Step 1: Inspect the DBP column

```python
print("DBP dtype:", df_raw['DBP'].dtype)
print("Missing values:", df_raw['DBP'].isna().sum())
print(df_raw['DBP'].describe())
```

Step 2: Find and Remove Out of Range values

```python
df_raw.loc[df_raw['DBP'] < 40, 'DBP'] = None
df_raw.loc[df_raw['DBP'] > 120, 'DBP'] = None

print("Missing values after removing outliers:", df_raw['DBP'].isna().sum())
```

Step 3: Replace missing values with median of the intial dataset to not skew the original median

```python
dbp_median = df_raw['DBP'].median()
print("DBP median:", dbp_median)

df_raw['DBP'] = df_raw['DBP'].fillna(dbp_median)
```

Step 4: Verification of Dataset

```python
print("Missing values after cleaning:", df_raw['DBP'].isna().sum())
print("Min:", df_raw['DBP'].min())
print("Max:", df_raw['DBP'].max())
print(df_raw['DBP'].describe())
```
