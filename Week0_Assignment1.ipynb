Step 8: Map Gender Column to Integers

```python
gender_map = {
    'Male'  : 1,
    'MALE'  : 1,
    '1'     : 1,
    'Female': 0,
    'FEMALE': 0,
    '0'     : 0
}

df_raw['gender_clean'] = df_raw['Gender'].map(gender_map)
```

Step 9: Check for Unmapped Values

```python
print("Unmapped values:", df_raw['gender_clean'].isna().sum())
```

Step 10: Drop Old Column and Rename Clean Column

```python
df_raw = df_raw.drop(columns=['Gender'])
df_raw = df_raw.rename(columns={'gender_clean': 'Gender'})
```
